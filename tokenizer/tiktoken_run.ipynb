```python
from tokenizer import Tiktokenizer

wrapped_tokenizer = Tiktokenizer()
```

```python
text = open('1984.txt').read()
```

```python
tokens = wrapped_tokenizer.encode(text, allowed_special={"<|endoftext|>", "<|bof|>"})
print(tokens)

# print the token to str conversions
wrapped_tokenizer.pretty_print_token_to_string(tokens)
```

```python
# Creating the input-output pairs
for input, output in wrapped_tokenizer.input_output_pairs(tokens):
    print(f'{input} -----> {output}')
```

```python
# Creating the input-output pairs, print the strings
for input, target in wrapped_tokenizer.input_output_pairs(tokens):
    print(f'{wrapped_tokenizer.decode(input)} -----> {wrapped_tokenizer.decode([target])}')
```
